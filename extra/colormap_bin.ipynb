#### This notebook generates a .bin file with c0lormap data for use by the Aggie XLMA (C++) application. 
Please generate the necessary file and update the code in state.cpp.
More colormaps can be found at [colorcet](https://colorcet.holoviz.org/), [cmocean](https://matplotlib.org/cmocean/), [cmasher](https://cmasher.readthedocs.io/index.html), and [crameri](https://www.fabiocrameri.ch/colourmaps/).

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
colormaps = ['viridis', 'plasma', 'inferno', 'magma', 'cividis']
t = np.linspace(0, 1, 256)
data = np.array([[plt.get_cmap(name)(v)[:3] for v in t] for name in colormaps], dtype=np.float32)
data.tofile('colormap.bin')
```
